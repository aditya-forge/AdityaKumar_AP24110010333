```python
# Using Helper Function For Sorting
def convert(obj):
    L = []
    for i in ast.literal_eval (obj):
        L.append(i['name'])
    return L
```

```python
import ast
ast.literal_eval('[{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}, {"id": 14, "name": "Fantasy"}, {"id": 878, "name": "Science Fiction"}]')
```

```python
import ast

movies['cast_x'] = movies['id'].map(
    credits.set_index('movie_id')['cast']
)

movies['crew_x'] = movies['id'].map(
    credits.set_index('movie_id')['crew']
)
```

```python
movies.head()
```

```python
def convert3(obj):
    L = []
    counter = 0
    for i in ast.literal_eval (obj):
        if counter != 3:
             L.append(i['name'])
             counter +=1
        else:
            break
    return L
```

```python
movies['cast_x']= movies['cast_x'].apply(convert3)
```

```python
movies.head()
```

```python
movies['crew_x'][0]
```

```python
def fetch_director(obj):
    L = []
    for i in ast.literal_eval (obj):
        if i in ast.literal_eval(obj):
            if i['job'] == 'Director':
                L.append(i['name'])
                break
    return L
```

```python
movies['crew_x'].apply(fetch_director)
```

```python
movies['crew_x'] = movies['crew_x'].apply(fetch_director)
```

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
```
